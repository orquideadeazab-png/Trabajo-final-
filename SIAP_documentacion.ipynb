# SIAP — Sistema Inteligente de Atención a la Diversidad Pedagógica
### Cuaderno de documentación y sustentación del proyecto final

**Laboratorio de programación en Python**

Este cuaderno documenta el desarrollo del proyecto **SIAP**, una aplicación construida con **Streamlit** que apoya a los docentes en la atención a la diversidad pedagógica de sus estudiantes, aplicando los conocimientos adquiridos durante el laboratorio (estructuras de datos, funciones, módulos propios, manejo de datos con pandas y construcción de una aplicación web funcional).


## 1. El problema o necesidad abordada

En un mismo salón de clases conviven estudiantes con:

- **Estilos de aprendizaje distintos** (visual, auditivo, kinestésico).
- **Ritmos de aprendizaje diferentes** (algunos avanzan rápido, otros necesitan más tiempo y repaso).
- **Necesidades específicas** (atención, lectoescritura, procesamiento matemático, comunicación, motricidad, altas capacidades, entre otras).

Atender esta diversidad de forma individual, sin herramientas de apoyo, es una tarea compleja y demandante para cualquier docente, especialmente en grupos numerosos. Muchas veces el diagnóstico de estilo de aprendizaje se hace de forma intuitiva y las estrategias pedagógicas no se diferencian realmente por estudiante.

**Necesidad concreta:** contar con una herramienta simple que ayude a diagnosticar el perfil de aprendizaje de cada estudiante y traduzca ese perfil en **recomendaciones pedagógicas accionables**, tanto a nivel individual como de grupo completo.


## 2. La solución desarrollada

**SIAP** es una aplicación web que:

1. Aplica un **cuestionario de 5 preguntas** basado en el modelo VAK (Visual - Auditivo - Kinestésico) a cada estudiante.
2. Calcula el **porcentaje de cada estilo** y el **estilo dominante**.
3. Registra el **ritmo de aprendizaje** observado y la **necesidad específica** reportada (si la hay).
4. Con esos tres factores (estilo, ritmo, necesidad), un **motor de recomendaciones basado en reglas** genera un **plan de atención a la diversidad** con estrategias concretas para ese estudiante.
5. Ofrece un **dashboard del aula** con gráficos de la distribución de estilos, ritmos y necesidades de todo el grupo, útil para planificar actividades diferenciadas a nivel de clase.
6. Permite **descargar** tanto el plan individual como el registro completo de estudiantes.

La lógica del motor de recomendaciones (módulo `src/recomendador.py`) está separada de la interfaz (`app.py`), siguiendo buenas prácticas de organización de código en Python (separación de responsabilidades, funciones puras y reutilizables, y una base de conocimiento declarada como datos y no “quemada” dentro de la lógica).


## 3. El framework seleccionado y por qué

Se seleccionó **Streamlit** como framework para construir la aplicación.

**Razones de la elección:**

- Permite construir una interfaz web interactiva usando **exclusivamente Python**, sin necesidad de HTML, CSS o JavaScript, lo cual encaja directamente con los conocimientos del laboratorio.
- Tiene un **ciclo de desarrollo muy rápido**: widgets como formularios, selectores y botones de descarga se crean con una sola línea de código.
- Se integra de forma nativa con **pandas** y **Plotly**, ideal para mostrar datos y gráficos del aula.
- Permite un **despliegue gratuito e inmediato** en Streamlit Community Cloud, cumpliendo el requisito de tener un enlace de despliegue funcional.
- Es ampliamente usado en proyectos de ciencia de datos y prototipado rápido de aplicaciones, lo que lo hace una herramienta valiosa para seguir aprendiendo más allá del laboratorio.


## 4. Principales funcionalidades de la aplicación

| Funcionalidad | Descripción |
|---|---|
| Diagnóstico VAK | Cuestionario de 5 preguntas que calcula el perfil Visual / Auditivo / Kinestésico de un estudiante |
| Registro de ritmo y necesidades | Captura el ritmo de aprendizaje y la necesidad específica reportada |
| Motor de recomendaciones | Genera estrategias pedagógicas combinando estilo, ritmo y necesidad |
| Plan de atención descargable | Exporta el plan individual en formato `.txt` |
| Dashboard del aula | Gráficos interactivos con la diversidad de todo el grupo |
| Registro histórico | Tabla y exportación `.csv` de todos los estudiantes registrados |

A continuación se demuestra el funcionamiento del **motor de recomendaciones**, que es el núcleo lógico de la aplicación, ejecutando directamente las funciones del módulo `src/recomendador.py` (el mismo código que usa `app.py`).


## 5. Demostración del funcionamiento

Se importa el módulo del proyecto y se simula el diagnóstico de un estudiante, tal como ocurre dentro de la app.

```python
import sys
sys.path.append("..")  # permite importar el paquete src/ del proyecto

from src.recomendador import (
    PREGUNTAS_VAK,
    calcular_estilo_dominante,
    generar_plan_atencion,
    plan_a_texto,
)

# Mostrar el cuestionario que responde cada estudiante dentro de la app
for i, item in enumerate(PREGUNTAS_VAK, start=1):
    print(f"{i}. {item['pregunta']}")
    for letra, texto in item["opciones"].items():
        print(f"   [{letra}] {texto}")
    print()
```

```python
# Simulación: respuestas de un estudiante al cuestionario VAK
respuestas_estudiante = ["K", "K", "V", "K", "A"]

perfil = calcular_estilo_dominante(respuestas_estudiante)
print("Perfil de aprendizaje calculado:")
print(perfil)
```

```python
# Generar el plan de atención a partir del perfil, el ritmo y la necesidad reportada
plan = generar_plan_atencion(
    nombre="Estudiante de ejemplo",
    perfil_vak=perfil,
    ritmo="Necesita más tiempo y repaso",
    necesidad="Dificultad de atención / concentración",
)

print(plan_a_texto(plan))
```

Como se observa, el motor de recomendaciones combina automáticamente:

- Estrategias según el **estilo dominante** (en este ejemplo, Kinestésico).
- Ajustes según el **ritmo de aprendizaje** (más tiempo y repaso).
- Ajustes según la **necesidad específica** (dificultad de atención).

Este mismo resultado es el que la aplicación Streamlit muestra en pantalla y permite descargar en la sección **📋 Recomendaciones y plan**, junto con el dashboard visual del grupo completo en **📊 Dashboard del aula**.
